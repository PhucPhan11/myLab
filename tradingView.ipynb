```python
from tvDatafeed import TvDatafeed, Interval

# 1. Khởi tạo kết nối (chế độ ẩn danh, không cần tài khoản)
tv = TvDatafeed()

# 2. Lấy dữ liệu 1 nến (phiên) gần nhất của cổ phiếu PVT trên sàn HOSE
df = tv.get_hist(symbol='PVT', exchange='HOSE', interval=Interval.in_daily, n_bars=1)

# 3. Trích xuất giá đóng cửa (close) của nến đó và in ra
gia_hien_tai = df['close'].iloc[0]
print(f"Giá PVT hiện tại: {gia_hien_tai:,.0f} VNĐ")
```
