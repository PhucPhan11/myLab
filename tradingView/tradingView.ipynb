# Stock Portfolio Analysis

This notebook tracks and analyzes a stock portfolio position. The analysis fetches real-time data from TradingView for the Vietnamese stock **BSR** (listed on **HOSE** exchange) and calculates key portfolio metrics including current valuation, profit/loss amount, and percentage gains/losses.

**Portfolio Summary:**
- **Symbol:** HOSE:BSR
- **Entry Price:** 33,200 VND
- **Current Price:** 30,600 VND
- **Position Size:** 30 shares
- **Current Status:** Loss of -7.83% (-78,000 VND)

The code demonstrates how to use TvDatafeed to retrieve historical stock data and perform fundamental portfolio calculations.

**BSR**

```python
from tvDatafeed import TvDatafeed, Interval

# Initialize TradingView connection
tv = TvDatafeed()

# Stock information
symbol = 'BSR'
exchange = 'HOSE'

# Your portfolio information
buy_price = 33200        # Buy price per share
quantity = 30          # Number of shares

# Fetch latest stock data
df = tv.get_hist(
    symbol=symbol,
    exchange=exchange,
    interval=Interval.in_daily,
    n_bars=1
)

# Get latest close price
current_price = df['close'].iloc[0]

# Calculate profit/loss percentage
profit_loss_percent = (
    (current_price - buy_price) / buy_price
) * 100

# Calculate actual profit/loss amount
profit_loss_amount = (
    current_price - buy_price
) * quantity

# Calculate portfolio values
investment_value = buy_price * quantity
current_value = current_price * quantity

# Display results
print("=" * 50)
print(f"Stock: {exchange}:{symbol}")
print("=" * 50)

print(f"Buy Price:        {buy_price:,.0f} VND")
print(f"Current Price:    {current_price:,.0f} VND")

print(f"Quantity:         {quantity:,} shares")

print(f"Initial Value:    {investment_value:,.0f} VND")
print(f"Current Value:    {current_value:,.0f} VND")

if profit_loss_percent >= 0:
    print(f"Profit:           +{profit_loss_percent:.2f}%")
    print(f"Profit Amount:    +{profit_loss_amount:,.0f} VND")
else:
    print(f"Loss:             {profit_loss_percent:.2f}%")
    print(f"Loss Amount:      {profit_loss_amount:,.0f} VND")

print("=" * 50)
```
